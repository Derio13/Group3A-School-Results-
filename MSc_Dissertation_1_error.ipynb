```python
!pip install xgboost -q

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier
```

Step 2: Upload and load your dataset

```python
from google.colab import files

uploaded = files.upload()

df = pd.read_csv("students_data.csv")

df.head()
```

Step 3: Prepare the data

```python
# Predictor variables
X = df.drop(columns=[
    "Student_ID",
    "target variable "
])

# Target variable
y = df["target variable "]

# Display class frequencies
print(y.value_counts())
```

```python
# Predictor variables (X)
X = df.drop(columns=[
    "Student_ID",
    "target variable "
])

# Target variable (y)
y = df["target variable "]

# Display the target classes
print(y.value_counts())
```

Step 4: One-hot encode all categorical variables

```python
categorical_cols = [
    "Gender",
    "District",
    "School_Type",
    "Parent_Education",
    "Internet_Access",
    "Private_Tuition"
]

X = pd.get_dummies(
    X,
    columns=categorical_cols,
    drop_first=True
)

print(X.shape)
X.head()
```

Step 5: Encode the target variable

```python
le = LabelEncoder()

y = le.fit_transform(y)

print("Class labels:")
print(le.classes_)
```

Step 6: Split into 80% training and 20% testing

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)
```

Objective 1: Decision Tree Classification

```python
dt = DecisionTreeClassifier(
    max_depth=6,
    random_state=42
)

dt.fit(X_train, y_train)

dt_pred = dt.predict(X_test)

print("Decision Tree Accuracy:",
      accuracy_score(y_test, dt_pred))

print(classification_report(
    y_test,
    dt_pred,
    target_names=le.classes_
))
```

```python
cm_dt = confusion_matrix(y_test, dt_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_dt,
    display_labels=le.classes_
)

disp.plot(cmap="Blues", values_format="d")
plt.title("Decision Tree Confusion Matrix")
plt.show()
```

Objective 2: Random Forest Classification

```python
rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf.fit(X_train, y_train)

rf_pred = rf.predict(X_test)

print("Random Forest Accuracy:",
      accuracy_score(y_test, rf_pred))

print(classification_report(
    y_test,
    rf_pred,
    target_names=le.classes_
))
```

random forest confusion matrix

```python
cm_rf = confusion_matrix(y_test, rf_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_rf,
    display_labels=le.classes_
)

disp.plot(cmap="Greens", values_format="d")
plt.title("Random Forest Confusion Matrix")
plt.show()
```

objective 3: Support Vector Machine (SVM)

```python
svm = SVC(
    kernel="rbf",
    C=1,
    gamma="scale"
)

svm.fit(X_train, y_train)

svm_pred = svm.predict(X_test)

print("SVM Accuracy:",
      accuracy_score(y_test, svm_pred))

print(classification_report(
    y_test,
    svm_pred,
    target_names=le.classes_
))
```

confusion matrix

```python
cm_svm = confusion_matrix(y_test, svm_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_svm,
    display_labels=le.classes_
)

disp.plot(cmap="Oranges", values_format="d")
plt.title("SVM Confusion Matrix")
plt.show()
```

objective 4: xgboost

```python
xgb = XGBClassifier(
    n_estimators=200,
    max_depth=4,
    learning_rate=0.1,
    random_state=42,
    eval_metric="mlogloss"
)

xgb.fit(X_train, y_train)

xgb_pred = xgb.predict(X_test)

print("XGBoost Accuracy:",
      accuracy_score(y_test, xgb_pred))

print(classification_report(
    y_test,
    xgb_pred,
    target_names=le.classes_
))
```

confusion matrix xgboost

```python
cm_xgb = confusion_matrix(y_test, xgb_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_xgb,
    display_labels=le.classes_
)

disp.plot(cmap="Purples", values_format="d")
plt.title("XGBoost Confusion Matrix")
plt.show()
```

Objective 5: Variable Importance Analysis

```python
importance = pd.DataFrame({
    "Variable": X.columns,
    "Importance": rf.feature_importances_
})

importance = importance.sort_values(
    "Importance",
    ascending=False
)

importance.head(13)
```

plot the variables

```python
top13= importance.head(13)

plt.figure(figsize=(8,6))

plt.barh(
    top13["Variable"],
    top13["Importance"]
)

plt.gca().invert_yaxis()

plt.xlabel("Importance Score")
plt.title("Top 13 Important Variables")

plt.show()
```

```python
results = pd.DataFrame({
    "Model": [
        "Decision Tree",
        "Random Forest",
        "Support Vector Machine",
        "XGBoost"
    ],
    "Accuracy": [
        accuracy_score(y_test, dt_pred),
        accuracy_score(y_test, rf_pred),
        accuracy_score(y_test, svm_pred),
        accuracy_score(y_test, xgb_pred)
    ]
})

results.sort_values("Accuracy", ascending=False)
```
